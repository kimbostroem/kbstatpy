# Demo 12 — Multiple dependent variables + correlation

**Dataset:** `iris` (R base / Fisher, 1936) — sepal and petal measurements for two iris species.  
**Question:** Do versicolor and virginica differ on all four morphological measurements, and how are the measurements correlated?  
**Model:** `<measurement> ~ Species`, one LM per outcome

## Setup

```python
import os
try:
    notebook_dir = os.path.dirname(os.path.abspath(__vsc_ipynb_file__))
except NameError:
    notebook_dir = os.getcwd()  # JupyterLab sets CWD to the notebook directory
os.chdir(notebook_dir)

import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions
```

## Options

`y` is a comma-separated list — kbstat runs the full pipeline independently for each variable. `correlation = options.y` adds a pairwise correlation analysis. `constraints` excludes setosa.

```python
options = KbstatOptions()
options.in_file     = os.path.join(notebook_dir, '../data/iris.csv')
options.out_dir     = os.path.join(notebook_dir, 'results/demo_12_multi_y')
options.y           = 'Sepal.Length, Sepal.Width, Petal.Length, Petal.Width'
options.y_units     = 'cm'
options.x           = 'Species'
options.constraints = 'Species != "setosa"'
options.correlation = options.y
options.rename      = 'Sepal.Length -> SepalLength; Sepal.Width -> SepalWidth; Petal.Length -> PetalLength; Petal.Width -> PetalWidth'
```

## Run

`run()` iterates over all four dependent variables and also runs the correlation analysis. Results are saved into per-variable subdirectories.

Results are saved automatically to `out_dir` at the end of `run()`.

```python
kb = Kbstat(options)
kb.run()
```

## Interpretation

- Four separate LMs are fitted — check each ANOVA and post-hoc table in the results subfolders.
- The correlation scatter grid shows how the four measurements relate to each other across both species combined.
- Petal measurements are expected to show stronger species differences than sepal measurements.
