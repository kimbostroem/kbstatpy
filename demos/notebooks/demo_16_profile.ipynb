# Demo 16 — Level-wise profile analysis (`profile_across`)

When one factor is an ordered series of levels, the question is often not "is there
an effect at some level?" but "how does another factor's effect change *across* the
ordered levels?" — the pattern is the finding. Here, on the `ToothGrowth` dataset:
odontoblast length of 60 guinea pigs given vitamin C as orange juice (`OJ`) or
ascorbic acid (`VC`) at three ordered doses (low, medium, high).

The model is the crossed two-way `len ~ supp * dose`. Setting
`profile_across = 'dose'` adds a level-wise profile of the **supp** effect across
the ordered dose levels:

- **Layer 1 (per level)** — the OJ-vs-VC contrast within each dose level.
- **Layer 2 (trend)** — the supp×dose interaction as a focused 1-df **linear trend**
  across the ordered dose positions, next to the factor-omnibus `supp:dose`.

The classic finding falls out: OJ's advantage over VC is large at low/medium dose
but vanishes at high dose — a monotone attenuation the linear trend captures.

## Setup

```python
import os

import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions
```

## Options

Same crossed model as Demo 3, plus `profile_across = 'dose'`. `x_order` fixes the
dose ordering so the linear trend runs low → medium → high. `out_dir` is empty, so
results (including the profile plot) render inline only.

```python
options = KbstatOptions()
options.in_file        = os.path.join(options.demo_dir, 'data/toothgrowth.csv')
options.out_dir        = ''    # empty: show results inline only; set a folder to also save them
options.y              = 'len'
options.y_units        = 'mm'
options.x              = 'supp, dose'
options.interaction    = 'supp, dose'    # needed for a non-flat profile
options.x_order        = 'dose: low, medium, high'
options.profile_across = 'dose'
```

## Run

`run()` fits the model and, because `profile_across` is set, adds the level-wise
profile: the per-dose supp contrast, the linear-trend vs factor-omnibus interaction
tests, and the profile plot — all rendered inline.

```python
kb = Kbstat(options)
kb.run();
```

## Interpretation

- The **profile plot** shows the OJ and VC curves across dose; the gap between them
  is the supp effect, and it narrows as dose rises.
- **Layer 1** quantifies that gap per dose (large and significant at low/medium,
  ~0 at high).
- **Layer 2** reports the supp×dose interaction two ways: the focused 1-df linear
  trend (the attenuation) alongside the factor-omnibus — the focused trend is the
  headline, per the "focused trend beats diffuse omnibus" logic.
- Set `out_dir` to also write `LevelProfile.xlsx` (per-level + trend tables) and
  `LevelProfile.pdf/png`.
