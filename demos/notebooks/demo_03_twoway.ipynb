# Demo 3 — Two-way ANOVA equivalent

This demo uses the `ToothGrowth` dataset (R base), in which the tooth
(odontoblast) length of 60 guinea pigs was measured after vitamin C
supplementation given by two delivery methods — orange juice (`OJ`) and ascorbic
acid (`VC`) — at three dose levels (low, medium, high), with 10 animals in each of
the six cells. It asks whether length depends on the supplement, the dose, and
their interaction.

Both factors enter as crossed between-subject effects:

    len ~ supp * dose

With effects coding, Type III sums of squares, and a balanced design, the result
is numerically identical to a classical two-way factorial ANOVA, including the
interaction term.

## Setup

```python
import os
try:
    notebook_dir = os.path.dirname(os.path.abspath(__vsc_ipynb_file__))
except NameError:
    notebook_dir = os.getcwd()  # JupyterLab sets CWD to the notebook directory
os.chdir(notebook_dir)

import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions
```

## Options

Two fixed-effect factors with an interaction term. No random effects — plain linear model equivalent to a classical two-way ANOVA.

```python
options = KbstatOptions()
options.in_file     = os.path.join(notebook_dir, '../data/toothgrowth.csv')
options.out_dir     = os.path.join(notebook_dir, 'results/demo_03_twoway')
options.y           = 'len'
options.y_units     = 'mm'
options.x           = 'supp, dose'
options.interaction = 'supp, dose'
options.rename      = 'len -> ToothLength; supp -> Supplement; dose -> Dose; supp: OJ -> orange_juice, VC -> vitamin_c'
options.x_order     = 'dose: low, medium, high'
```

## Model fitting

`fit()` estimates the model parameters via restricted maximum likelihood (REML).

```python
kb = Kbstat(options)
kb.fit()
print(f'Formula : {kb._build_formula()}')
print(f'AIC     : {kb.AIC:.3f}')
print(f'BIC     : {kb.BIC:.3f}')
print(f'logLik  : {kb.logLik:.3f}')
```

## ANOVA table (Type III)

```python
kb.anova()
kb.anova_table
```

## Post-hoc pairwise comparisons

Pairwise comparisons for each factor. Check whether the interaction is significant before interpreting main effects.

```python
kb.posthoc()
kb.posthoc_table
```

## Data plot

Violin + jitter scatter with EMM and 95 % CI overlaid.

> An interactive version with hover tooltips is saved to `results/` as an HTML file.

```python
kb.plot_data()
plt.show()
```

## Diagnostic plots

Six panels checking model assumptions.

See [STATISTICAL_NOTES.md](../../STATISTICAL_NOTES.md) for interpretation guidance.

```python
kb.plot_diagnostics()
plt.show()
```

## Save results

Writes all output files to `out_dir`.

```python
kb.save()
```

## Interpretation

- If the interaction (Supplement × Dose) is significant, main effects should be interpreted with caution.
- The data plot shows one panel per supplement type with dose on the x-axis.
- Post-hoc comparisons are computed for the first factor (`supp`); examine the data plot for the interaction pattern.
