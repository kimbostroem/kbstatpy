# Demo 9 — LMM with partial interaction

**Dataset:** `npk` (R base) — nitrogen, phosphate, and potassium effects on pea yield.  
**Question:** Do N, P, and K affect yield, with only the P×K interaction hypothesised?

## Setup

```python
import os
try:
    notebook_dir = os.path.dirname(os.path.abspath(__vsc_ipynb_file__))
except NameError:
    notebook_dir = os.getcwd()  # JupyterLab sets CWD to the notebook directory
os.chdir(notebook_dir)

import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions
```

## Options

Three binary treatment factors. `interaction = 'P, K'` tests only the P×K interaction — N is modelled as a main effect only. Formula: `yield ~ N + P*K + (1 | block)`.

```python
options = KbstatOptions()
options.in_file     = os.path.join(notebook_dir, '../data/npk.csv')
options.out_dir     = os.path.join(notebook_dir, 'results/demo_09_lmm_partial_interaction')
options.y           = 'yield'
options.y_units     = 'lb/plot'
options.x           = 'N, P, K'
options.id          = 'block'
options.interaction = 'P, K'
options.rename      = ('yield -> CropYield; N -> Nitrogen; P -> Phosphate; K -> Potassium; block -> Block; '
                       'N: 0 -> absent, 1 -> applied; P: 0 -> absent, 1 -> applied; K: 0 -> absent, 1 -> applied')
options.x_order     = 'Nitrogen: absent, applied; Phosphate: absent, applied; Potassium: absent, applied'
```

## Model fitting

`fit()` estimates the model parameters via restricted maximum likelihood (REML).

```python
kb = Kbstat(options)
kb.fit()
print(f'Formula : {kb._build_formula()}')
print(f'AIC     : {kb.AIC:.3f}')
print(f'BIC     : {kb.BIC:.3f}')
print(f'logLik  : {kb.logLik:.3f}')
```

## ANOVA table (Type III)

```python
kb.anova()
kb.anova_table
```

## Post-hoc pairwise comparisons

Pairwise comparisons for the primary factor (Nitrogen). Check the ANOVA table for the P×K interaction row.

```python
kb.posthoc()
kb.posthoc_table
```

## Data plot

Violin + jitter scatter with EMM and 95 % CI overlaid.

> An interactive version with hover tooltips is saved to `results/` as an HTML file.

```python
kb.plot_data()
plt.show()
```

## Diagnostic plots

Six panels checking model assumptions.

See [STATISTICAL_NOTES.md](../../STATISTICAL_NOTES.md) for interpretation guidance.

```python
kb.plot_diagnostics()
plt.show()
```

## Save results

Writes all output files to `out_dir`.

```python
kb.save()
```

## Interpretation

- Specifying a partial interaction reflects a domain hypothesis: N is not expected to interact with P or K.
- If the P×K interaction is significant, the effect of phosphate depends on whether potassium is applied.
- Compare the model AIC with a fully-saturated model (all interactions) to assess whether the partial structure is justified.
