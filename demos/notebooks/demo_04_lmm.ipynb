# Demo 4 — One-way repeated-measures ANOVA (LMM)

**Dataset:** `ergoStool` (nlme) — perceived effort to arise from 4 stool types, rated by 9 subjects.  
**Question:** Do stool types differ in perceived effort?

## Setup

```python
import os
try:
    notebook_dir = os.path.dirname(os.path.abspath(__vsc_ipynb_file__))
except NameError:
    notebook_dir = os.getcwd()  # JupyterLab sets CWD to the notebook directory
os.chdir(notebook_dir)

import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions
```

## Options

One within-subject factor (`Type`) with a random intercept per subject. Balanced design — equivalent to a classical one-way repeated-measures ANOVA.

```python
options = KbstatOptions()
options.in_file  = os.path.join(notebook_dir, '../data/ergostool.csv')
options.out_dir  = os.path.join(notebook_dir, 'results/demo_04_lmm')
options.y        = 'effort'
options.y_units  = 'Borg'
options.x        = 'Type'
options.id       = 'Subject'
options.rename   = 'effort -> Effort; Type -> StoolType'
```

## Model fitting

`fit()` estimates the model parameters via restricted maximum likelihood (REML).

```python
kb = Kbstat(options)
kb.fit()
print(f'Formula : {kb._build_formula()}')
print(f'AIC     : {kb.AIC:.3f}')
print(f'BIC     : {kb.BIC:.3f}')
print(f'logLik  : {kb.logLik:.3f}')
```

## ANOVA table (Type III)

```python
kb.anova()
kb.anova_table
```

## Post-hoc pairwise comparisons

Four stool types — six pairwise comparisons with Holm correction.

```python
kb.posthoc()
kb.posthoc_table
```

## Data plot

Violin + jitter scatter with EMM and 95 % CI overlaid.

> An interactive version with hover tooltips is saved to `results/` as an HTML file.

```python
kb.plot_data()
plt.show()
```

## Diagnostic plots

Six panels checking model assumptions.

See [STATISTICAL_NOTES.md](../../STATISTICAL_NOTES.md) for interpretation guidance.

```python
kb.plot_diagnostics()
plt.show()
```

## Save results

Writes all output files to `out_dir`.

```python
kb.save()
```

## Interpretation

- The random intercept absorbs between-subject baseline differences in perceived effort.
- For a balanced design with a single random intercept, this is an exact one-way RM-ANOVA equivalent.
- The post-hoc table identifies which specific stool types differ significantly.
