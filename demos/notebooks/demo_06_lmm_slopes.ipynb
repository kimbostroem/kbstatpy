# Demo 6 — LMM with random intercepts and random slopes

Subjects often differ not just in their baseline but in how strongly a treatment
affects them — something repeated-measures ANOVA cannot express, but random slopes
can. This demo shows that on the `sleepstudy` dataset (lme4; Belenky et al. 2003):
reaction times of 18 subjects across a period of sleep deprivation (here a rested
block versus a deprived one).

Allowing each subject their own slope as well as their own intercept gives:

    Reaction ~ Period + (1 + Period | Subject)

This has no classical equivalent — RM-ANOVA assumes a single shared time effect.
The random-effects table now carries an additional slope variance and an
intercept–slope correlation.

## Setup

```python
import os
try:
    notebook_dir = os.path.dirname(os.path.abspath(__vsc_ipynb_file__))
except NameError:
    notebook_dir = os.getcwd()  # JupyterLab sets CWD to the notebook directory
os.chdir(notebook_dir)

import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions
```

## Options

`slope = 'Period'` adds a random slope: each subject gets their own Period effect in addition to their own baseline. Compare with Demo 7 which uses the same dataset with a log transform instead.

```python
options = KbstatOptions()
options.in_file  = os.path.join(notebook_dir, '../data/sleepstudy.csv')
options.out_dir  = ''   # empty: show results inline only; set a folder to also save them
options.y        = 'Reaction'
options.y_units  = 'ms'
options.x        = 'Period'
options.id       = 'Subject'
options.slope    = 'Period'
options.rename   = 'Reaction -> ReactionTime; Period -> Day'
```

## Run

`run()` executes the whole pipeline — fit, ANOVA, post-hoc comparisons, plots, and a printed summary — and writes files only if `out_dir` is set (empty here, so results are shown inline).

```python
kb = Kbstat(options)
kb.run()
```

## Interpretation

- The random effects table includes both intercept variance (between-subject baseline differences) and slope variance (between-subject differences in deprivation sensitivity).
- A positive intercept-slope correlation means subjects with higher baseline reaction times also show larger deprivation effects.
