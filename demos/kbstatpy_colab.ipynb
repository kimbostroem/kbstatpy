## Run any demo — here, in this session

Set `demo` below and run the cell. It loads that demo's configuration and runs it
with `run()`, so the tables and figures render **inline** (nothing written to
disk) in the kernel you already set up — **no new runtime and no re-install**.

```
demo_01_unpaired        demo_06_lmm_slopes             demo_11_glmm_binomial

```

See the [README demo table](https://github.com/kimbostroem/kbstatpy#demo-scripts)
for what each one shows.


````python
import os
from kbstatpy import KbstatOptions

demo = 'demo_01_unpaired'        # <- change to any demo name from the list above

# Load the demo's script and run it inline, swapping run_save() -> run() so it
# renders tables + figures here without writing files. Runs in THIS session.
path = os.path.join(KbstatOptions().demo_dir, 'scripts', demo + '.py')
exec(open(path).read().replace('run_save()', 'run()'))
````
