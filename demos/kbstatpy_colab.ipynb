## Inline example — Demo 1 (unpaired *t*-test equivalent)

A linear model with a single two-level factor reproduces the classical
independent-samples *t*-test exactly, on the classic `sleep` dataset. Running it
here renders the ANOVA table, post-hoc comparison, and data plot inline.

To try a different dataset, point `in_file` at any file in `demos/data/`
(`toothgrowth.csv`, `iris.csv`, `sleepstudy.csv`, …) and adjust `y` / `x`.


````python
import os
import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions

options = KbstatOptions()
options.in_file = os.path.join(options.demo_dir, 'data/sleep.csv')
options.out_dir = ''            # '' -> show inline only; set a folder to also save files
options.y       = 'extra'
options.y_units = 'h'
options.x       = 'group'
options.rename  = 'extra -> ExtraSleep; group -> DrugGroup'

kb = Kbstat(options)
kb.run()                        # fit + ANOVA + post-hoc + plots, rendered inline
````
