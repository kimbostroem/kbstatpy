## Setup 1/2 — R packages

Colab already has R and `rpy2`; this cell just adds the CRAN packages kbstatpy
relies on, pulling precompiled binaries from Posit Package Manager so it is fast.
Run it once (~2–4 min). It is safe to re-run — packages already present are skipped.


````python
%%bash
# Install the R packages kbstatpy uses. The Ubuntu codename is detected so the
# right precompiled binaries are fetched (falls back to jammy = Ubuntu 22.04).
Rscript -e '
codename <- tryCatch(system("lsb_release -cs", intern = TRUE), error = function(e) "jammy")
if (length(codename) == 0 || codename == "") codename <- "jammy"
options(repos = c(CRAN = sprintf("https://packagemanager.posit.co/cran/__linux__/%s/latest", codename)))
pkgs <- c("lme4", "lmerTest", "glmmTMB", "emmeans", "pbkrtest", "DHARMa",
          "tibble", "broom", "broom.mixed", "report", "see", "parameters",
          "performance", "effectsize", "insight", "datawizard", "bayestestR")
missing <- pkgs[!pkgs %in% rownames(installed.packages())]
if (length(missing)) {
    cat("Installing:", paste(missing, collapse = ", "), "\n")
    install.packages(missing)
} else {
    cat("All R packages already present.\n")
}
'
````

## Inline example — Demo 1 (unpaired *t*-test equivalent)

A linear model with a single two-level factor reproduces the classical
independent-samples *t*-test exactly, on the classic `sleep` dataset. Running it
here renders the ANOVA table, post-hoc comparison, and data plot inline.

To try a different dataset, point `in_file` at any file in `demos/data/`
(`toothgrowth.csv`, `iris.csv`, `sleepstudy.csv`, …) and adjust `y` / `x`.


````python
import os
import matplotlib.pyplot as plt
from kbstatpy import Kbstat, KbstatOptions

options = KbstatOptions()
options.in_file = os.path.join(options.demo_dir, 'data/sleep.csv')
options.out_dir = ''            # '' -> show inline only; set a folder to also save files
options.y       = 'extra'
options.y_units = 'h'
options.x       = 'group'
options.rename  = 'extra -> ExtraSleep; group -> DrugGroup'

kb = Kbstat(options)
kb.run()                        # fit + ANOVA + post-hoc + plots, rendered inline
````
